```python
import copernicusmarine

copernicusmarine.subset(
  dataset_id="cmems_obs-wind_glo_phy_my_l4_0.125deg_PT1H",
  variables=["eastward_wind", "northward_wind"],
  minimum_longitude=-179.9375,
  maximum_longitude=179.9375,
  minimum_latitude=66.56,
  maximum_latitude=90,
  start_datetime="2010-01-01T00:00:00",
  end_datetime="2010-01-31T23:59:00",
  output_filename="wind_2010-01.nc"
)
```

```python
import xarray as xr
import numpy as np
from pathlib import Path 
```

```python
ds = xr.open_dataset("wind_2010-01.nc")
ds
```

```python
encoding = {"eastward_wind": {"zlib": True, "complevel": 4}, "northward_wind": {"zlib": True, "complevel": 4}}
ds_east6 = ds["eastward_wind"].resample(time="6h").mean()
ds_north6 = ds["northward_wind"].resample(time="6h").mean()
ds_east6
ds_north6
```

```python
ws=xr.merge([ds_east6, ds_north6])
ws.to_netcdf("winds6hs_2010-01.nc", encoding=encoding)
```
